```python
#Import the libaries 
import pandas as pd 
import numpy as np
```

```python
#Load the Data files 
Goverment_spending=pd.read_csv("/Users/jahaankachwala/Downloads/Projects /education-spending/government-spending-on-education-as-a-share-of-gdp.csv")
Eu_students=pd.read_csv("/Users/jahaankachwala/Downloads/Projects /education-spending/european_university_students.csv")
international_student=pd.read_csv("/Users/jahaankachwala/Downloads/Projects /education-spending/european_international_students_quickstart.csv")
```

```python
#File Cleaning for goverment spending 
Goverment_spending = Goverment_spending.rename(columns={
    "Entity": "country",
    "Code": "country_code",
    "Total across all levels of education": "spending_pct_gdp",
})
Goverment_spending = Goverment_spending.drop_duplicates()
Goverment_spending["country"] = Goverment_spending["country"].str.strip()
Goverment_spending["Year"] = Goverment_spending["Year"].astype(int)
Goverment_spending = Goverment_spending.rename(columns={"Year": "year"})
Goverment_spending = Goverment_spending.dropna(subset=["spending_pct_gdp"])
```

```python
#File Cleaning for EU university students 
Eu_students = Eu_students.drop_duplicates()
Eu_students["date"] = pd.to_datetime(Eu_students["date"], errors="coerce")
Eu_students["country"] = Eu_students["country"].str.strip()
# remove aggregate rows (e.g. EU27) so countries aren't double counted
Eu_students = Eu_students[Eu_students["is_country_group"] == False]
Eu_students = Eu_students.dropna(subset=["students"])
 
```

```python
 #File Cleaning for International students
international_student = international_student.drop_duplicates()
international_student["date"] = pd.to_datetime(international_student["date"], errors="coerce")
international_student["country"] = international_student["country"].str.strip()
international_student = international_student.dropna(subset=["international_students"])
```

```python
#Checking results 
print("\nAfter cleaning:")
print("Goverment_spending:", Goverment_spending.shape)
print("Eu_students:", Eu_students.shape)
print("international_student:", international_student.shape)
```

```python
#Save cleaned files
out = "/Users/jahaankachwala/Downloads/Projects /education-spending/"
Goverment_spending.to_csv(out + "clean_government_spending.csv", index=False)
Eu_students.to_csv(out + "clean_eu_students.csv", index=False)
international_student.to_csv(out + "clean_international_students.csv", index=False)
```
